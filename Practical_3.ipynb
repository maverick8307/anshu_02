```python
import numpy as np

# 1. Universal Functions (ufuncs) & Vectorized Operations
x = np.array([1, 4, 9])
y = np.array([10, 20, 30])

print("Element-wise Add (np.add):", np.add(x, y))
print("Square Root (np.sqrt):     ", np.sqrt(x))
print("Exponential (np.exp):      ", np.exp(np.array([1, 2])))

# 2. Broadcasting Features (Operating on different shapes)
matrix = np.array([[10, 20, 30], 
                   [40, 50, 60]])  # Shape (2, 3)

# Scalar Broadcasting: 5 is stretched to match the (2, 3) matrix
print("\nMatrix * Scalar:\n", matrix * 5)

# Row Vector Broadcasting: Row (1, 3) stretches vertically down the rows
row_vec = np.array([1, 2, 3])      # Shape (3,)
print("\nMatrix + Row Vector:\n", matrix + row_vec)
```
